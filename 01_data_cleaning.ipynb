## 6. Clean the Movie–Genres Relationship Dataset


```python
print("Shape:", movie_genres.shape)

print("\nData types:")
print(movie_genres.dtypes)

print("\nMissing values:")
print(movie_genres.isnull().sum())

print("\nDuplicate rows:", movie_genres.duplicated().sum())

print("\nDuplicate movie-genre pairs:",
      movie_genres.duplicated(
          subset=["movie_id", "genre_id"]
      ).sum())
```

```python
movie_genres = movie_genres.drop_duplicates(
    subset=["movie_id", "genre_id"]
).copy()

print("Shape after cleaning:", movie_genres.shape)
print("Remaining duplicate rows:", movie_genres.duplicated().sum())
```

```python
invalid_movie_ids = movie_genres.loc[
    ~movie_genres["movie_id"].isin(movies["movie_id"])
]

invalid_genre_ids = movie_genres.loc[
    ~movie_genres["genre_id"].isin(genres["genre_id"])
]

print("Invalid movie IDs:", len(invalid_movie_ids))
print("Invalid genre IDs:", len(invalid_genre_ids))
```

```python
# Remove crew records that reference movies absent from the movies table.
crew = crew[crew["movie_id"].isin(movies["movie_id"])].copy()
print("Crew rows after orphan removal:", len(crew))
```

```python
# Remove relationship rows that point to missing movies or genres.
movie_genres = movie_genres[
    movie_genres["movie_id"].isin(movies["movie_id"])
    & movie_genres["genre_id"].isin(genres["genre_id"])
].copy()
print("Movie-genre rows after orphan removal:", len(movie_genres))
```

```python
movie_genres.to_csv("movie_genres_cleaned.csv", index=False)

print("movie_genres_cleaned.csv saved successfully!")
print("Final shape:", movie_genres.shape)
```

## 13. Financial Data Checks: Profit and ROI


```python
invalid_movies = movies[
    (movies["budget"].isna()) |
    (movies["revenue"].isna()) |
    (movies["budget"] <= 0) |
    (movies["revenue"] <= 0)
].copy()

print("Movies with missing/zero budget or revenue:",
      len(invalid_movies))

display(
    invalid_movies[
        ["title", "release_date", "budget", "revenue"]
    ].head(20)
)
```

```python
valid_movies = movies[
    (movies["budget"] > 0) &
    (movies["revenue"] > 0)
].copy()

valid_movies["profit"] = (
    valid_movies["revenue"] - valid_movies["budget"]
)

valid_movies["roi"] = (
    (valid_movies["revenue"] - valid_movies["budget"])
    / valid_movies["budget"]
) * 100

print("Movies with valid budget and revenue:", len(valid_movies))

display(
    valid_movies[
        ["title", "budget", "revenue", "profit", "roi"]
    ].head(10)
)
```

```python
top_15_roi = valid_movies[
    valid_movies["budget"] >= 1_000_000
].nlargest(15, "roi")

display(
    top_15_roi[
        ["title", "budget", "revenue", "profit", "roi"]
    ].reset_index(drop=True)
)
```

## 14. Release-Year and Runtime Analysis


```python
movies["release_year"] = pd.to_datetime(
    movies["release_date"], errors="coerce"
).dt.year

movies_per_year = (
    movies.groupby("release_year")["movie_id"]
    .nunique()
    .sort_values(ascending=False)
)

print("Year with highest movie output:")
print(movies_per_year.head(1))

print("\nTop 10 years by movie count:")
display(movies_per_year.head(10).reset_index(
    name="movie_count"
))
```

```python
Q1 = movies["runtime"].quantile(0.25)
Q3 = movies["runtime"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

runtime_outliers = movies[
    (movies["runtime"] < lower_bound) |
    (movies["runtime"] > upper_bound)
][["title", "runtime", "release_date"]]

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of runtime outliers:", len(runtime_outliers))

display(runtime_outliers.sort_values("runtime"))
```

## 15. Genre and Cast Analysis


```python
movie_genres = pd.read_csv("movie_genres_cleaned.csv")

primary_genres = (
    movie_genres
    .drop_duplicates(subset="movie_id", keep="first")
    .rename(columns={"genre_id": "primary_genre_id"})
)

primary_genres = primary_genres.merge(
    genres,
    left_on="primary_genre_id",
    right_on="genre_id",
    how="left"
)

genre_popularity = primary_genres.merge(
    movies[["movie_id", "title", "popularity"]],
    on="movie_id",
    how="inner"
)

avg_popularity = (
    genre_popularity
    .groupby("genre_name")["popularity"]
    .mean()
    .sort_values(ascending=False)
    .round(2)
)

display(avg_popularity.reset_index(name="avg_popularity"))
```

```python
cast = pd.read_csv("cast_cleaned.csv")

actor_movies = cast.merge(
    movies[["movie_id", "vote_average"]],
    on="movie_id",
    how="inner"
)

top_10_actors = (
    actor_movies
    .groupby(["person_id", "actor_name"])
    .agg(
        movie_count=("movie_id", "nunique"),
        avg_rating=("vote_average", "mean")
    )
    .sort_values("movie_count", ascending=False)
    .head(10)
    .reset_index()
)

top_10_actors["avg_rating"] = (
    top_10_actors["avg_rating"].round(2)
)

display(top_10_actors)
```

## 16. Duplicate Titles and Audience Engagement


```python
duplicate_titles = movies[
    movies.duplicated(subset="title", keep=False)
][["title", "release_date", "movie_id"]].copy()

duplicate_titles["release_year"] = pd.to_datetime(
    duplicate_titles["release_date"],
    errors="coerce"
).dt.year

duplicate_titles = duplicate_titles.sort_values(
    ["title", "release_year"]
)

print("Number of rows with duplicate titles:",
      len(duplicate_titles))

display(duplicate_titles)
```

```python
import numpy as np

movies["popularity_z"] = (
    movies["popularity"] - movies["popularity"].mean()
) / movies["popularity"].std()

movies["vote_count_z"] = (
    movies["vote_count"] - movies["vote_count"].mean()
) / movies["vote_count"].std()

anomalies = movies[
    ((movies["popularity_z"] > 2) & (movies["vote_count_z"] < 0)) |
    ((movies["vote_count_z"] > 2) & (movies["popularity_z"] < 0))
][[
    "title", "popularity", "vote_count",
    "popularity_z", "vote_count_z"
]].copy()

print("Number of unusual movies:", len(anomalies))

display(anomalies.sort_values(
    "popularity_z", ascending=False
))
```
